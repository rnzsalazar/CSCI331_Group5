# Query 1 #

Create a query using the rank window function. We want to use the Production.TransactionHistory to view transactions and the standings of their difference in actual cost. 
Using rank will allow users to see the difference

```sql

Select TransactionID,ProductID, Quantity, ActualCost,
	RANK() OVER(Order By ActualCost) as Rank
From Production.TransactionHistory
Where ActualCost > 0
```

Create a query that will show the total sales quota by first the city and the CountryRegion Name. Use Rollup to create this and Sales.vSalesPerson view instead of a table. We want to see the total for each country.

```sql
Select CountryRegionName, City, SUM(SalesQuota) AS TotalQuota
From Sales.vSalesPerson
Group By ROLLUP (CountryRegionName, City)
```

# Query 3 #
Create a query that will go through the subtotals of each day in the Sales.SalesOrderHeader table. This must be in a times series manner and will compare the current days subtotal to the previous on a different column.

```sql

;WITH DailySubtotal As (
SELECT
		CAST(OrderDate AS DATE) AS OrderPerDay,
		SUM(SubTotal) AS DailySubtotal
	From Sales.SalesOrderHeader
	Group BY CAST(OrderDate AS DATE)
)
Select 
	OrderPerDay, DailySubTotal,
	LAG(DailySubTotal, 1) OVER (Order BY OrderPerDay) AS PreviousSubtotal
From DailySubtotal
```

# Query 6 #
Create a query that is similar to Date bucket (Since SSMS does not support Date bucket), and create a table that summarizes the date for the 3 month fiscal starting at the first recorded data date, It should also provide the amount of transactions per the 3 months.

```sql
Select
	DATE_BUCKET(MONTH, 3, TransactionDate, '2013-07-31') AS FiscalQuarter,
	COUNT(*) AS TranscationID
From Production.TransactionHistory
Group By DATE_BUCKET(MONTH, 3, TransactionDate, '2013-07-31')


Select
	DATEADD (month, (DATEDIFF(month, '2013-07-31', TransactionDate) / 3) * 3, '2013-07-31') AS ThreeMonthFiscal,
	COUNT(*) AS TransactionCounts
From Production.TransactionHistory
GROUP BY DATEADD (month, (DATEDIFF(month, '2013-07-31', TransactionDate) / 3) * 3, '2013-07-31')
ORDER bY ThreeMonthFiscal
```

# Query 8 #
Create a query that will use user created grouping sets. We will want to use the account number, purchase order numers and the sum of the frieghts. Do not use a cube function. As well instread of grouping set index, create a custom name for the grouping sets based of trheir index.

```sql
Select

CASE
	WHEN GROUPING_ID([AccountNumber], [PurchaseOrderNumber]) = 3 THEN 'Grand Total'
	WHEN GROUPING_ID([AccountNumber], [PurchaseOrderNumber]) = 1 THEN 'AccountNumber'
	WHEN GROUPING_ID([AccountNumber], [PurchaseOrderNumber]) = 0 THEN 'AccountNumber, PurchaseOrderNumber'
	END AS 'Grouping Level'


	 ,[AccountNumber],
	[PurchaseOrderNumber],
	SUM(Freight) AS FrieghtTotal
From Sales.SalesOrderHeader
Group By GROUPING SETS
(
	(),
	([AccountNumber]),
	([AccountNumber], [PurchaseOrderNumber])
)
ORDER BY GROUPING_ID ([AccountNumber], [PurchaseOrderNumber]) DESC
```

# Query 9 #
 Create a query that will work such as the date bucket function, but this time we will want to see the analysis of the 7 work timeframe. This time we want to see the sum of the total due needed for the order

```sql
Select
	DATEADD(day, (DateDiff(Day, '2011-06-07', ShipDate) / 7 ) * 7, '2011-06-07') AS SevenDayAnalysis,
	SUM(TotalDue) AS SevenDaySales
From Sales.SalesOrderHeader
GROUP BY DATEADD(day, (DateDiff(Day, '2011-06-07', ShipDate) / 7 ) * 7, '2011-06-07')
ORDER BY SevenDayAnalysis
```
