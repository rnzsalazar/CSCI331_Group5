### Fully Qualified Domains within Taxonomy ###
Fullu Qualified Domain Name is essentially the address to a server. The FQDN includes all parts of a domain which leads the hostname down right to the root domain. These are important since this is the method to how computers can find and connect to services. In SQL, it is important to provide the complete address to ensure that the computer can reach the database that encases all the objects. Taxonomy is the practice of organizing data in a structured hierarchy and has many different subcategories. The FQDN ensures that the correct database is connected to which helps prevent collisions if different database have same schemas or tables. This way, we are able to keep data consistant since we are concretely hitting the same database and again able to use the same data in the database. The same goes for control of the data, knowing that we are having concrete knowledge that we are working on a specific server is essential to control security and policies. 

### Importance of Fully Qualified Table Names ###
Fully qualified table names is referring to the creation of table inside a database and schema. This would make these tables very specific to the schema and its own database. So if another schema or database has the same table name, there would be no confusion since they are in different "Classes".
Essentially, it is a method that is used to avoid collision of names. Each time someone has to reference a specific table, there are safety points to make sure the correct table is selected. We can also ensure that scalability can be handled since we have such specific paths to tables and schemas. So if large sums of data are inputted, we know that they are not being scattered across wrong tables and creating unnecessary conflicts
