CREATE TYPE and CREATE DOMAIN

Chances are that certain fields will be repeated multiple times across different tables in a database. Creating a custom datatype can help keep the same type of data consistent across all your tables by defining a base datatype, fixed length and any type of constraints or formatting for that value. This ensures that the same rules are applied throughout the database to improve data consistency.

SQL Server uses CREATE TYPE to create your own named versions of an existing data type. 
For example, if you had a column that consists of zip codes, you need to enforce the fact they consist of a maximum of 5 characters like so:

CREATE TYPE ZipCode
FROM CHAR(5)
NOT NULL;

We also make it a CHAR and not an INT because there are zip codes that have leading zeros, and using an INT would drop those zeros.
Another example is if we are storing social security numbers:

CREATE TYPE SSN
FROM CHAR(11)
NOT NULL;

We set a max number of characters at 11 because social security numbers are made up of 9 numbers and 2 dashes. 
Let’s say we have an Employee, Customer or Warehouse table, a field that they share among each other would be a Phone Number or a State Code, so these are great examples that we can use for our custom datatypes to keep the data formatting the same for all of these tables.
So we can just create the 2 data types like so:

CREATE TYPE StateCode
FROM CHAR(2)
NOT NULL;

And

CREATE TYPE PhoneNumber
FROM VARCHAR(12)
NOT NULL;

Then these can be used for the Customer and Employee table when defining a State or CellPhone column. 


ANSI Standard’s CREATE DOMAIN is similar to CREATE TYPE, in that you can create aliases for built in data types. They do have an additional part which is a CHECK constraint, which allows you to enforce a rule on the data. 

CREATE DOMAIN TEST_DOM DATE
DEFAULT NULL
CHECK (VALUE > DATE'1900-01-01' OR
VALUE IS NULL)
NOT DEFERRABLE;

In this example, it verifies that any non NULL values being inserted have to be later than January 1, 1900. 

So CREATE TYPE and CREATE DOMAIN are similar in that you can define alias for your own data types you want to create. However, in a CREATE DOMAIN you can use a CHECK constraint to immediately validate the values that belong to that domain. While if you use CREATE TYPE, you have to use the CHECK constraint when using a CREATE TABLE to validate the values for that specific column you are creating.
