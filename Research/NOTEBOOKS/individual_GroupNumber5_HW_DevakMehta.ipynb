## 1. What is a Taxonomy in a Database
A Taxonomy is a hierarchical classification system that organizes entities, attributes, and relationships to create structure and semantic meaning across a database ecosystem(DAMA,2017) 

```python
-- taxonomy-based schema heirarchy
CREATE SCHEMA Sales; --Creates a schema named Sales. In taxonomy terms, this is a top-level node that “owns” sales objects.
CREATE SCHEMA Production; --Taxonomy node for manufacturing/operations.
CREATE SCHEMA HumanResources; --Taxonomy node for people/HR data

--each has its taxonomy node
CREATE TABLE Sales.Customers ( ----Defines a customer entity inside the Sales taxonomy node, giving you the FQTN Sales.Customer.
    CustomerID INT NOT NULL PRIMARY KEY, -- Integer surrogate key; PRIMARY KEY enforces uniqueness + index; NOT NULL ensures every row has an ID.
    CustomerName VARCHAR(100), -- Unicode name up to 100 chars; required (NOT NULL).
    ContactEmail VARCHAR(100) --Unicode email up to 255; optional at this table level (we’ll later enforce a reusable email domain).

);

CREATE TABLE Production.Products ( -- Product entity under Production taxonomy → Production.Product (its FQTN).
    ProductID INT PRIMARY KEY, -- Primary key for products
    Category NVARCHAR(50),
    Price 50

);

```

## How it connects:
Each Schema is a node in the taxonomy, a clear classification of where data belongs

**Citation:** DAMA International. (2017). The DAMA Guide to the Data Management Body of Knowledge (2nd ed.). Technics Publications.

## 2. Taxonomy & Data Governance 
Taxonomies provide controlled terms for metadata, buisness rules, and lineage. IBM calls this taxonomy-driven Governance (IBM,2023)

```python
CREATE TYPE EMAILADDRESSTYPE FROM NVARCHAR(100) NOT NULL;

CREATE TABLE HR.Employees (
    EmployeeID INT PRIMARY KEY,
    EmployeeName VARCHAR(100),
    ContactEmail EMAILADDRESSTYPE
);
-- shows consistency in data types across schemas. This ensures all team use the same format for email addresses, promoting data integrity/quality across the organization.
```

## 3. Taxonomies + Fully Qualified Domains (FQDs) 
**What are FQDs?** "A Fully Qualified Domain is a domain that includes all labels, including the top-level domain and the root"
 (RFC 1035,https://datatracker.ietf.org/doc/html/rfc1035#section-3.1)


FQDs extend the taxonomy principle into resuable data types


```python
CREATE TYPE Sales.EmailAddressType FROM NVARCHAR(100);

--FQD: schema.object type
--Sales.EmailAddressType is the fully qualified domain for email addresses in the Sales schema
--this is similar to CREATE DOMAIN in the ANSI SQL standard, now the use of taxonomy classifies or clears where this domain belongs and what semantic meaning it carries. 
--(citations: ISO/IEC 9075-2:2016. Information technology—Database languages—SQL—Part 2: Foundation.)
```

## 4. Taxonomies + Fully Qualified Table Names (FQTNs)
FQTNs provide explicit reference paths that map directly to taxonomy levels improving clarity and scalibility

```python
--reference table with its FQTN
SELECT s.CustomerName, s.ContactEmail
FROM Sales.Customers AS s
JOIN Sales.Orders AS o ON s.CustomerID = o.CustomerID;

--using Sales.Customer rather than just customer removes ambiguity, enhances clarity, and aligns with taxonomy levels for better data governance.
-- citations:Microsoft. (2024). Database Identifiers (Transact-SQL).
--https://learn.microsoft.com/en-us/sql/relational-databases/databases/database-identifiers
```

## 5. Real-World Example (AdventureWorks)
The AdventureWorks sample databse uses schemas as a taxonomy:
    - used Sales, Production, etc
    - each schema holds related tables with reusable data types.
    - Each schema holds its own related tables and domain types.
    -For example:
        1. Production.Product manages product data,
        2. Sales.Customer handles customers, and
        3. HumanResources.Employee tracks personnel.
        That division makes it immediately clear which department owns which data and simplifies security, maintenance, and reporting.
